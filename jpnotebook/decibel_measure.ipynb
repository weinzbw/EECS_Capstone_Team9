```python
"""
Program Name: Decibel.py
Description: finds decibels
Programmer(s): Sam Harrison, ChatGPT
Date Made: 4/22
Date(s) Revised:
Preconditions: 
Postconditions: 
Errors/Exceptions:
Side Effects:
Invariants: 
Known Faults:
"""

import numpy as np
from scipy.io import wavfile

# Load the WAV file
sample_rate, data = wavfile.read("test.wav")

# Normalize the data to [-1, 1] range if it's in integer format
if data.dtype == np.int16:
    data = data / 32768.0
elif data.dtype == np.int32:
    data = data / 2147483648.0
elif data.dtype == np.uint8:
    data = (data - 128) / 128.0

# If stereo, just use one channel
if len(data.shape) > 1:
    data = data[:, 0]

# Calculate RMS (root mean square)
rms = np.sqrt(np.mean(data**2))

# Convert to decibels
db = 20 * np.log10(rms) if rms > 0 else -np.inf

print(f"Decibel level: {db:.2f} dB")
```
