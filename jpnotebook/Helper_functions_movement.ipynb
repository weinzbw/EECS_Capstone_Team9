```python
from pynq.overlays.base import BaseOverlay
from pynq.lib.pmod import Pmod_IO
import time

# Load the base overlay to access GPIO pins
base = BaseOverlay("base.bit")

# Define the Pmod_IO pins for motor control (connected to L298N)
motor1_in1 = Pmod_IO(base.PMODA, 4, "out")  # IN1 pin (Motor 1) -> IO4
motor1_in2 = Pmod_IO(base.PMODA, 5, "out")  # IN2 pin (Motor 1) -> IO5
motor2_in1 = Pmod_IO(base.PMODA, 6, "out")  # IN3 pin (Motor 2) -> IO6
motor2_in2 = Pmod_IO(base.PMODA, 7, "out")  # IN4 pin (Motor 2) -> IO7

motor1_in3 = Pmod_IO(base.PMODA, 0, "out")  # IN1 pin (Motor 1) -> IO0
motor1_in4 = Pmod_IO(base.PMODA, 1, "out")  # IN2 pin (Motor 1) -> IO1
motor2_in3 = Pmod_IO(base.PMODA, 2, "out")  # IN3 pin (Motor 2) -> IO2
motor2_in4 = Pmod_IO(base.PMODA, 3, "out")  # IN4 pin (Motor 2) -> IO3
```

```python
# Function to move forward
def moveForward(duration):
    print("Moving forward")
    motor1_in1.write(1)  # Motor 1 - IN1 = High
    motor1_in2.write(0)  # Motor 1 - IN2 = Low
    motor2_in1.write(1)  # Motor 2 - IN3 = High
    motor2_in2.write(0)  # Motor 2 - IN4 = Low
    motor1_in3.write(1)  # Motor 1 - IN1 = High
    motor1_in4.write(0)  # Motor 1 - IN2 = Low
    motor2_in3.write(1)  # Motor 2 - IN3 = High
    motor2_in4.write(0)  # Motor 2 - IN4 = Low
    time.sleep(duration)  # Run motors for the specified duration
    stopMotors()  # Stop the motors after the duration

# Function to move backward
def moveBackward(duration):
    print("Moving backward")
    motor1_in1.write(0)  # Motor 1 - IN1 = Low
    motor1_in2.write(1)  # Motor 1 - IN2 = High
    motor2_in1.write(0)  # Motor 2 - IN3 = Low
    motor2_in2.write(1)  # Motor 2 - IN4 = High
    motor1_in3.write(0)  # Motor 1 - IN1 = Low
    motor1_in4.write(1)  # Motor 1 - IN2 = High
    motor2_in3.write(0)  # Motor 2 - IN3 = Low
    motor2_in4.write(1)  # Motor 2 - IN4 = High
    time.sleep(duration)  # Run motors for the specified duration
    stopMotors()  # Stop the motors after the duration

# Function to turn right
def turnRight(duration):
    print("Turning right")
    motor1_in1.write(1)  # Motor 1 - IN1 = High
    motor1_in2.write(0)  # Motor 1 - IN2 = Low
    motor2_in1.write(0)  # Motor 2 - IN3 = Low
    motor2_in2.write(1)  # Motor 2 - IN4 = High
    motor1_in3.write(1)  # Motor 1 - IN1 = High
    motor1_in4.write(0)  # Motor 1 - IN2 = Low
    motor2_in3.write(0)  # Motor 2 - IN3 = Low
    motor2_in4.write(1)  # Motor 2 - IN4 = High
    time.sleep(duration)  # Turn for the specified duration
    stopMotors()  # Stop the motors after the duration

# Function to turn left
def turnLeft(duration):
    print("Turning left")
    motor1_in1.write(0)  # Motor 1 - IN1 = Low
    motor1_in2.write(1)  # Motor 1 - IN2 = High
    motor2_in1.write(1)  # Motor 2 - IN3 = High
    motor2_in2.write(0)  # Motor 2 - IN4 = Low
    motor1_in3.write(0)  # Motor 1 - IN1 = Low
    motor1_in4.write(1)  # Motor 1 - IN2 = High
    motor2_in3.write(1)  # Motor 2 - IN3 = High
    motor2_in4.write(0)  # Motor 2 - IN4 = Low
    time.sleep(duration)  # Turn for the specified duration
    stopMotors()  # Stop the motors after the duration

# Function to stop the motors
def stopMotors():
    print("Motors stopped")
    motor1_in1.write(0)  # Motor 1 - IN1 = Low
    motor1_in2.write(0)  # Motor 1 - IN2 = Low
    motor2_in1.write(0)  # Motor 2 - IN3 = Low
    motor2_in2.write(0)  # Motor 2 - IN4 = Low
    motor1_in3.write(0)  # Motor 1 - IN1 = Low
    motor1_in4.write(0)  # Motor 1 - IN2 = Low
    motor2_in3.write(0)  # Motor 2 - IN3 = Low
    motor2_in4.write(0)  # Motor 2 - IN4 = Low
```
