```python
from pynq.lib.arduino import Arduino_IO
import time

motor1_in1 = Arduino_IO(2, 'out')  # fl
motor1_in2 = Arduino_IO(3, 'out')  # bl
motor2_in1 = Arduino_IO(4, 'out')  # fr
motor2_in2 = Arduino_IO(5, 'out')  # br

def move_forward():
    motor1_in1.write(1)
    motor1_in2.write(0)
    motor2_in1.write(1)
    motor2_in2.write(0)

def turn_left():
    motor1_in1.write(0) # left wheel goes backwards
    motor1_in2.write(1)
    motor2_in1.write(1) # right wheel goes forwards
    motor2_in2.write(0)

def turn_right():
    motor1_in1.write(1) # left wheel goes forwards
    motor1_in2.write(0)
    motor2_in1.write(0) # right wheel goes backwards
    motor2_in2.write(1)

def stop_motors():
    motor1_in1.write(0)
    motor1_in2.write(0)
    motor2_in1.write(0)
    motor2_in2.write(0)
```
