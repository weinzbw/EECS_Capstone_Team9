```python
"""
Program Name: BackwardsButton.py
Description: It reverses the wheel movement, going backwards by clicking button 0
Programmer(s): Sam Harrison
Date Made: 4/6
Date(s) Revised:
Preconditions: 
Postconditions: 
Errors/Exceptions:
Side Effects:
Invariants: 
Known Faults:
"""

from pynq.overlays.base import BaseOverlay
from pynq.lib.pmod import Pmod_IO
from pynq import GPIO
import time

base = BaseOverlay("base.bit")

motor1_in1 = Pmod_IO(base.PMODA, 4, "out")  # IN1 pin (Motor 1) -> IO4 
motor1_in2 = Pmod_IO(base.PMODA, 5, "out")  # IN2 pin (Motor 1) -> IO5
motor2_in1 = Pmod_IO(base.PMODA, 6, "out")  # IN3 pin (Motor 2) -> IO6
motor2_in2 = Pmod_IO(base.PMODA, 7, "out")  # IN4 pin (Motor 2) -> IO7

motor1_in3 = Pmod_IO(base.PMODA, 0, "out")  # IN1 pin (Motor 1) -> IO0
motor1_in4 = Pmod_IO(base.PMODA, 1, "out")  # IN2 pin (Motor 1) -> IO1
motor2_in3 = Pmod_IO(base.PMODA, 2, "out")  # IN3 pin (Motor 2) -> IO2
motor2_in4 = Pmod_IO(base.PMODA, 3, "out")  # IN4 pin (Motor 2) -> IO3

print("Motor started")

while base.buttons[0].read() == 0:
    pass

motor1_in1.write(0)
motor1_in2.write(1)
motor1_in3.write(0)
motor1_in4.write(1)

motor2_in1.write(0)
motor2_in2.write(1)
motor2_in3.write(0)
motor2_in4.write(1)

time.sleep(1)

motor1_in1.write(0)
motor1_in2.write(0)
motor2_in1.write(0)
motor2_in2.write(0)

motor1_in3.write(0)
motor1_in4.write(0)
motor2_in3.write(0)
motor2_in4.write(0)

```
